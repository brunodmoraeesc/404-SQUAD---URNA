```python
import pandas as pd
import numpy as np

COLUNAS_UT = [2, 5, 9, 11, 12, 13, 16, 18, 26, 27, 28, 29]

pt = pd.read_csv(
    'p_turno.csv',
    encoding='latin1',
    sep=';',
    usecols=COLUNAS_UT
)
st = pd.read_csv(
    's_turno.csv',
    encoding='latin1',
    sep=';',
    usecols=COLUNAS_UT
)
df = pd.concat([pt, st], axis=0, ignore_index=True)
df = df.rename(columns={
    "SG_ UF" : "UF",
    "NM_MUNICIPIO" : "MUNICIPIO",
    "NR_TURNO" : "TURNO",
    "NR_ZONA" : "ZONA",
    "DS_CARGO_PERGUNTA" : "CARGO",
    "ANO_ELEICAO" : "ANO",
    "DS_TIPO_VOTAVEL" : "TIPO_VOTO",
    "NR_VOTAVEL" : "NR_VOTADO" ,
    "NM_VOTAVEL" : "NOME_VOTADO"
})

df.to_csv('bases_juntas_cols.csv')

display(df.info())
```

```python
def formata_string(coluna, case='title'):
    coluna = coluna.astype(str).str.replace(r'\s+', ' ', regex=True).str.strip()
    return coluna.str.upper() if case == 'upper' else coluna.str.title()

def formatar_inteiro(coluna):
    return pd.to_numeric(coluna, errors='coerce').fillna(0).astype(int)

cols_upper = ["UF", "SG_PARTIDO", "MUNICIPIO"]
cols_title = ["CARGO", "NOME_VOTADO"]
cols_int = ["QT_VOTOS"]

df[cols_upper] = df[cols_upper].apply(lambda c: formata_string(c, case='upper'))
df[cols_title] = df[cols_title].apply(formata_string)
df[cols_int] = df[cols_int].apply(formatar_inteiro)

display(df)
```
